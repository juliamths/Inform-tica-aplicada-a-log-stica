```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sakshisatre/ice-cream-sales-dataset")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Localizar o arquivo CSV no caminho do dataset baixado
files = os.listdir(path)
csv_files = [f for f in files if f.endswith('.csv')]

if csv_files:
    file_path = os.path.join(path, csv_files[0])
    df = pd.read_csv(file_path)
    print("Colunas encontradas no dataset:", df.columns.tolist())

    # Tentando mapear as colunas de faturamento (x) e fatura/temperatura (y)
    # Ajuste os nomes das colunas conforme o seu arquivo se necessário
    x_col = df.columns[1] # Geralmente a segunda coluna (ex: faturamento/vendas)
    y_col = df.columns[0] # Geralmente a primeira coluna (ex: temperatura/fatura)

    print(f"Plotando Gráfico de Dispersão: {x_col} (Eixo X) vs {y_col} (Eixo Y)")

    # Criar o gráfico de dispersão
    plt.figure(figsize=(8, 6))
    plt.scatter(df[x_col], df[y_col], color='blue', alpha=0.7, edgecolors='k')
    plt.title('Gráfico de Dispersão')
    plt.xlabel('Faturamento (x)')
    plt.ylabel('Fatura (y)')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()
else:
    print("Nenhum arquivo CSV encontrado na pasta do dataset.")
```
