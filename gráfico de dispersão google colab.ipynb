```python
import pandas as pd

# URL do arquivo de dados brutos (raw)
url = 'https://raw.githubusercontent.com/Sainathmudigonda/Student-Result-Analysis/main/student_exam_scores.csv'

# Importando os dados
df = pd.read_csv(url)

# Exibindo as primeiras linhas do DataFrame
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.linear_model import LinearRegression

# Preparando os dados
X = df[['hours_studied']]
y = df['exam_score']

# Treinando o modelo de regressão linear
model = LinearRegression()
model.fit(X, y)

# Obtendo os coeficientes
coef_angular = model.coef_[0]
intercepto = model.intercept_
r_quadrado = model.score(X, y)

# Gerando as previsões para a linha de regressão
y_pred = model.predict(X)

# Criando o gráfico
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='hours_studied', y='exam_score', alpha=0.7, label='Dados Reais')
plt.plot(df['hours_studied'], y_pred, color='red', linewidth=2, label='Linha de Regressão')

# Adicionando títulos e legenda
plt.title('Relação entre Horas Estudadas e Nota do Exame', fontsize=14)
plt.xlabel('Horas Estudadas (x)', fontsize=12)
plt.ylabel('Nota do Exame (y)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Exibindo os valores dos coeficientes no gráfico
texto_info = f'Equação: y = {coef_angular:.2f}x + {intercepto:.2f}\n$R^2$ = {r_quadrado:.4f}'
plt.gca().text(0.05, 0.95, texto_info, transform=plt.gca().transAxes, fontsize=12,
             verticalalignment='top', bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

plt.show()

# Imprimindo os coeficientes no console
print(f"Coeficiente Angular (Inclinação): {coef_angular:.4f}")
print(f"Intercepto (Intersecção com o eixo Y): {intercepto:.4f}")
print(f"Coeficiente de Determinação (R²): {r_quadrado:.4f}")
```
